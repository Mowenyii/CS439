## Linear regression

Last week, logistic regression helped us predict a class. Linear regression is different: it predicts a number.

We want to find a line that fits our data. We can use this line to make predictions:

$$\hat y = \theta_0 + \theta_1x$$

Here, $x$ is the input and $\hat y$ is the prediction. $\theta_0$ moves the line up or down, and $\theta_1$ controls the slope.

So, does linear regression predict a class or a number? **A number.**


## How do we check our predictions?

Of course, our line won't be perfect. We can compare each prediction with the real value.

One way to measure the error is **mean squared error**, or **MSE**:

$$\mathrm{MSE} = \frac{1}{N}\sum_{i=1}^{N}(\hat y_i-y_i)^2$$

We square the errors and take the average. A smaller MSE means the line fits our data better.

Let's look at a few data points.


```python
import numpy as np
import matplotlib.pyplot as plt


x = np.array([0., 1., 2., 3., 4.])
y = np.array([0., 2., 0., 5., 4.])

# Show the data
plt.scatter(x, y, label="Real data")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

## How do we improve the line?

We start with a line, check its loss, and then try to make that loss smaller.

This is where **gradient descent** comes in. A derivative tells us how the loss changes when a parameter changes. We use that information to update the parameters:

$$\theta \leftarrow \theta - \alpha\nabla L(\theta)$$

Here, $\alpha$ is the **learning rate**. It controls how big each update is.

In practice, many machine learning tools can calculate derivatives automatically.


## Python demo

Let's run this example. We start with a simple line, and the code updates it many times.


```python
# Ready-to-run instructor demo (no live coding needed)
def f(theta0, theta1, x):
    return theta0 + theta1 * x

def L(theta0, theta1, x, y):
    return np.mean((f(theta0, theta1, x) - y) ** 2)

# Derivatives of MSE (the original notebook's two gradients)
def L_0(theta0, theta1, x, y):
    return np.mean(2 * (f(theta0, theta1, x) - y))

def L_1(theta0, theta1, x, y):
    return np.mean(2 * (f(theta0, theta1, x) - y) * x)

theta0, theta1 = 0.0, 0.0  # Simple starting line to see a clear improvement
print("Before training - MSE:", round(L(theta0, theta1, x, y), 3))

start_line = f(theta0, theta1, x)
alpha = 0.01
for _ in range(1000):
    # Compute both gradients BEFORE changing parameters
    d0 = L_0(theta0, theta1, x, y)
    d1 = L_1(theta0, theta1, x, y)
    theta0 -= alpha * d0
    theta1 -= alpha * d1

print("After training  - MSE:", round(L(theta0, theta1, x, y), 3))

plt.scatter(x, y, label="Real data")
plt.plot(x, start_line, '--', label="Before")
plt.plot(x, f(theta0, theta1, x), label="After")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

Here is the loss before training, and here is the loss after training. It's smaller now.

And here are the two lines. You can see how the line changed to fit the data better.
