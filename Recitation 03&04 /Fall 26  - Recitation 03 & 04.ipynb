### Aggregation (input function is: column series -> value)
https://stackoverflow.com/questions/27517425/apply-vs-transform-on-a-group-object

```python
df_grouped = df.groupby('Product Type')
display_all_groups(df_grouped)
```

```python
import numpy as np
#---------------------------------------------
#* single function *
# compute sum of each group
display(df_grouped.sum())
display(df_grouped.agg(func=sum))
display(df_grouped.agg(func='sum'))
display(df_grouped.agg(func=np.sum))

# compute size of each group
display(df_grouped.size())
display(df_grouped.agg(func='size'))

# compute describe of each group
display(df_grouped.describe())
display(df_grouped.agg(func='describe'))

# custom function, lambda function
display(df_grouped.agg(func=np.max))
display(df_grouped.agg(func=np.min))
```

```python
# Answer
def agg_f(column):
    #print(group)
    return np.max(column) - np.min(column)

display(df_grouped['Age'].agg(func=agg_f))
```

```python
#* multiple functions *
display(df_grouped['Age'].agg(func=[np.sum, np.mean]))
```

```python
#* different function for different column *
display(df_grouped.agg(func={'Age': np.sum, 'Sale Price': np.mean}))
```

```python
# Named Aggregation: https://pandas.pydata.org/pandas-docs/stable/user_guide/groupby.html#named-aggregation
display(df_grouped.agg(sum_age=('Age', np.sum), mean_price=('Sale Price', np.mean)))
```

### Concatenate

```python
d1 = df.loc[:3]
d2 = df.loc[3:]
display(d1)
display(d2)
```

```python
pd.concat([d1, d2], axis=0)
# With axis=0, pandas stacks the DataFrames vertically, so we're adding rows.
```

```python
left = pd.DataFrame({'A': ['A0', 'A1', 'A2'],
                      'B': ['B0', 'B1', 'B2']},
                     index=['K0', 'K1', 'K2'])
right = pd.DataFrame({'C': ['C0', 'C2', 'C3'],
                       'D': ['D0', 'D2', 'D3']},
                      index=['K0', 'K2', 'K3'])
display(left)
display(right)
left.join(right)
# join aligns the two DataFrames using their indices.
```

### merge

```python
d1 = df.loc[:3, ['Contact', 'Age', 'Product ID']]
d2 = df.loc[3:, ['Contact', 'Age', 'Product ID']]
display(d1)
display(d2)
```

```python
# We're merging the two DataFrames using Contact as the key.
merged_df = d1.merge(right=d2, how='left', on="Contact")
merged_df
# The left table decides which keys must stay.
# Pandas looks for a matching Contact value on the right.
# If it finds one, the values are combined.
# If it doesn't find one, the columns coming from the right side contain NaN.
```

```python
merged_df = d1.merge(right=d2, how='right', on="Contact")
merged_df
```

```python
merged_df = d1.merge(right=d2, how='inner', on="Contact")
merged_df
# An inner merge keeps only keys that appear in both DataFrames.
```

```python
merged_df = d1.merge(right=d2, how='outer', on="Contact")
merged_df
#An outer merge keeps all keys from both DataFrames.
```

```python
merged_df = d1.merge(right=d2, how='outer', on=["Contact", "Product ID"])
merged_df
# The pair of columns together acts as the key.
```
